```python
import rustbpe
```

```python
text = open('../tiny-shakespear.txt', 'r').read()
text[:100]
```

```python
tokeniser = rustbpe.Tokenizer()
```

```python
tokeniser.train_from_iterator([text], vocab_size=4096)
print(f"Trained Vocab Size: {tokeniser.vocab_size}")
```

```python
sample_text = "To be, or not to be, that is the question:"
tokens = tokeniser.encode(sample_text)
decoded = tokeniser.decode(tokens)

print(f"Encoded tokens ({len(tokens)}): {tokens}")
print(f"Decoded: {decoded}")
```

```python
token_ids = tokeniser.encode(sample_text)

# 2. Get the index-to-bytes mapping
# get_mergeable_ranks() returns Dict[bytes, int] mapping raw bytes to token IDs
vocab_map = {v: k for k, v in tokeniser.get_mergeable_ranks()}

# 3. View the list of individual tokens as string representations
tokens_as_bytes = [vocab_map[idx] for idx in token_ids]
tokens_as_strings = [b.decode('utf-8', errors='replace') for b in tokens_as_bytes]

print("Token IDs: ", token_ids)
print("Token Strings: ", tokens_as_strings)
# Example output: ['ROM', 'EO', ':', ' To', ' be', ',', ' or', ' not', ' to', ' be', '!']

# Zip them together to view ID -> Subword alignment directly
for idx, token_str in zip(token_ids, tokens_as_strings):
    print(f"ID {idx:>5} -> {repr(token_str)}")
```
